```python
import matplotlib.pyplot as plt

# Titik A
A = (3, 3)

# Transformasi refleksi terhadap titik O(0,0)
def refleksi_O(point):
    x, y = point
    return (-x, -y)

# Titik hasil
A1 = refleksi_O(A)

# Plot
plt.figure(figsize=(6,6))

# Plot titik A dan A'
plt.scatter(A[0], A[1])
plt.scatter(A1[0], A1[1])
plt.scatter(0, 0)  # titik O

# Garis penghubung
plt.plot([A[0], A1[0]], [A[1], A1[1]], linestyle='--')

# Garis bantu (seperti di gambar)
plt.plot([A[0], A[0]], [0, A[1]], linestyle='dashed')
plt.plot([0, A[0]], [A[1], A[1]], linestyle='dashed')

plt.plot([A1[0], A1[0]], [0, A1[1]], linestyle='dashed')
plt.plot([0, A1[0]], [A1[1], A1[1]], linestyle='dashed')

# Label titik
plt.text(A[0], A[1], " A(x,y)")
plt.text(A1[0], A1[1], " A'(x',y')")
plt.text(0, 0, " O")

# Sumbu
plt.axhline(0)
plt.axvline(0)

# Grid & batas
plt.grid(True)
plt.xlim(-5, 5)
plt.ylim(-5, 5)

plt.title("Refleksi Titik terhadap Titik O(0,0)")
plt.show()

# Output koordinat
print("Titik A :", A)
print("Titik A':", A1)
```
